```python
import torch
from torchvision.transforms import v2
import PIL
from PIL import Image
import os
import numpy as np
from pathlib import Path

torch.manual_seed(95)
```

```python
in_file = 'Data/NIH_Chest_XRay/images/00000001_000.png'
img_nih = Image.open('Data/NIH_Chest_XRay/images/00000001_000.png')
```

```python
type(img_nih)
print(img_nih.size, '\n', img_nih.mode)
```

```python
transform = v2.Compose([
        v2.ToImage(),
        # v2.Resize((resize_to, resize_to), antialias=True),
        v2.ToDtype(torch.float32, scale=True), 
    ])
```

```python
img_nih = transform(img_nih)
```

```python
img_r = img_nih * 12000
```

```python
img_p = torch.poisson(img_r)
img_p = img_p / 12000
```

```python
torch.mean((img_nih == img_p) * 1.0)
```

```python
transform_to_image = v2.Compose([
        v2.ToDtype(torch.uint8, scale=True),
        # v2.Grayscale(num_output_channels=1),
        v2.ToPILImage(),
    ])
```

```python
img_poisson = transform_to_image(img_p)
```
